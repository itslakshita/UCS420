```python
import pandas as pd
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25",
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5],
}

df = pd.DataFrame(data)

df.to_csv("employees.csv", index=False)

df3 = pd.read_csv("employees.csv")
```

```python
avg_salary = df["Salary"].mean()
print(f"Average Salary: {avg_salary}")
```

```python
total_bonus = df["Bonus"].sum()
print(f" Total Bonus: {total_bonus}")
```

```python
youngest_age = df["Age"].min()
print(f" Youngest Employee's Age: {youngest_age}")
```

```python
highest_rating = df["Rating"].max()
print(f" Highest Performance Rating: {highest_rating}")
```

```python
df_sorted = df.sort_values(by="Salary", ascending=False)
print(df_sorted)
```

```python
import numpy as np
conditions = [
    df['Rating'] >= 4.5,
    (df['Rating'] >= 4.0) & (df['Rating'] < 4.5)
]
choices = ['Excellent', 'Good']
df['Performance_Category'] = np.select(conditions, choices, default='Average')
print(df[['Employee_ID', 'Name', 'Rating', 'Performance_Category']])
```

```python
df.isnull().sum()
```

```python
df.rename(columns={'Employee_ID': 'ID'}, inplace=True)
```

```python
experienced_employees = df[df['Years_of_Experience'] > 5]
print(experienced_employees)
```

```python
IT_employees = df[df['Department'] == "IT"]
print(IT_employees)
```

```python
df['Tax'] = df['Salary'] * 0.10
print(df[['ID', 'Name', 'Salary', 'Tax']])
```

```python
df.to_csv('modified_employees.csv', index=False)
```
