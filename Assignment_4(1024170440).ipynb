#Question 2

```python
def search_and_score(query, df):

    query_words = query.lower().split()
    scores = []

    for _, row in df.iterrows():
        score = 0
        q_text = row["question"].lower()
        k_text = row["keywords"].lower()
        c_text = row["category"].lower()

        for word in query_words:
            if word in k_text:
                score += 2
            if word in q_text:
                score += 1
            if word in c_text:
                score += 1

        if query.lower() in q_text:
            score += 5

        scores.append(score)

    results = df.copy()
    results["confidence_score"] = scores

    ranked_results = results[results["confidence_score"] > 0].sort_values(
        by="confidence_score", ascending=False
    )

    return ranked_results
```

```python
def search_and_score(query, df):
    query_words = query.lower().split()
    scores = []


    for _, row in df.iterrows():
        score = 0
        q_text = row["question"].lower()
        k_text = row["keywords"].lower()
        c_text = row["category"].lower()

        for word in query_words:
            if word in k_text:
                score += 2
            if word in q_text:
                score += 1
            if word in c_text:
                score += 1

        if query.lower() in q_text:
            score += 5

        scores.append(score)

    results = df.copy()
    results["confidence_score"] = scores


    matching_results = results[results["confidence_score"] > 0]

    if matching_results.empty:
        print(f"No matches found for query: '{query}'")
        return matching_results


    max_score = matching_results["confidence_score"].max()


    top_matches = matching_results[
        matching_results["confidence_score"] == max_score
    ]

    if len(top_matches) > 1:
        print(
            f"Query '{query}' produced a TIE between {len(top_matches)} entries (Score: {max_score}):"
        )
    else:
        print(
            f"Query '{query}' produced a SINGLE top match (Score: {max_score}):"
        )

    return top_matches[["question", "answer", "confidence_score"]]
```
