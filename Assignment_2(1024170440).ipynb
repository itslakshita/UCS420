# Question 3

```python
import random
random.seed(1024170440)
```

```python
numbers = [random.randint(100,900) for i in range (100)]
print(numbers )
```

```python
odd_numbers = [num for num in numbers if num % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Number of odd numbers:", len(odd_numbers))
```

```python
even_numbers = [num for num in numbers if num % 2 == 0]

print("Odd numbers:", even_numbers)
print("Number of odd numbers:", len(even_numbers))
```

```python
def prime(num):
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
prime_numbers = [num for num in numbers if prime(num)]

print("Prime numbers:", prime_numbers)
print("Count:", len(prime_numbers))
```

```python
most_frequent = max(numbers, key=numbers.count)

print("Most frequent number:", most_frequent)
print("Number of times it occurs:", numbers.count(most_frequent))
```
