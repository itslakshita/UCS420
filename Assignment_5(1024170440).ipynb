```python
arr = np.array([1,2,3,4,5])
reverse = arr[::-1]
print(reverse)
```

```python
def find_most_frequent(arr):
  vals, counts = np.unique(arr,return_counts= True)
  max_count = counts.max()
  most_frequent = vals[counts == max_count]
  for val in most_frequent:
   indices = np.where(arr == val )[0]
   print(f"Value: {val}, Indices: {indices}")

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
find_most_frequent(x)
```

```python
import numpy as np
Lakshita = np.linspace(10,100,25)
print("Array:\n", Lakshita)
print("Dimensions (ndim):", Lakshita.ndim)
print("Shape:", Lakshita.shape)
print("Total elements (size):", Lakshita.size)
print("Data type (dtype):", Lakshita.dtype)
print("Total bytes consumed (nbytes):", Lakshita.nbytes)


Lakshita_transposed = Lakshita.reshape(-1, 1)
print("\nTransposed using reshape(-1, 1):\n", Lakshita_transposed)

print("\nUsing .T attribute on 1D array:\n", Lakshita.T)
```

#Question 5


```python
import numpy as np

ucs420_lakshita = np.array(
    [[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]]
)

print("Original Array:\n", ucs420_lakshita)
print("Mean:", np.mean(ucs420_lakshita))
print("Median:", np.median(ucs420_lakshita))
print("Max:", np.max(ucs420_lakshita))
print("Min:", np.min(ucs420_lakshita))
print("Unique Elements:", np.unique(ucs420_lakshita))

reshaped_ucs420_lakshita = ucs420_lakshita.reshape(4, 3)
print("\nReshaped Array (4x3):\n", reshaped_ucs420_lakshita)

resized_ucs420_lakshita = np.resize(ucs420_lakshita, (2, 3))
print("\nResized Array (2x3):\n", resized_ucs420_lakshita)
```
